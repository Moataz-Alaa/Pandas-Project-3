# Data Cleaning
1) Rename the columns using clear and readable names.
2) Handle missing values when necessary.
3) Remove duplicate rows.
4) Remove unnecessary columns.
5) Make sure all columns have the correct data types.
6) Sort the data where appropriate.

8) Create a new column called satisfied based on rating.
Satisfied rule: rating >= 7 → Satisfied Else Not Satisfied

1) Rename the columns using clear and readable names.

```python
df.columns = df.columns.str.strip()
```

2) Handle missing values when necessary.

```python
df.isnull().sum()
```

```python
df['Unit price'] = df['Unit price'].fillna(df['cogs'] / df['Quantity'])
df['Sales'] = df['Sales'].str.replace('EGP', '').astype('float')
df['Sales'] = df['Sales'].fillna(df['cogs'] * 1.05)
df = df.dropna(subset=['Date'])
```

```python
df.isnull().sum()
```

3) Remove duplicate rows.

```python
df = df.drop_duplicates()
```

4) Remove unnecessary columns.

```python
df = df.drop(columns=['Tax 5%', 'gross income', 'gross margin percentage'])
df
```

5) Make sure all columns have the correct data types.

```python
df.info()
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df['Branch'] = df['Branch'].str.strip().str.lower()
df['Payment'] = df['Payment'].str.strip().str.lower()

cat_cols = ['Branch', 'City', 'Customer type', 'Gender', 'Product line', 'Payment']

df[cat_cols] = df[cat_cols].astype('category')
```

```python
df.info()
```

6) Sort the data where appropriate.

7) Create a new 3 columns Day, Month, Year

```python
df['day'] = df['Date'].dt.day
df['month'] = df['Date'].dt.month
df['year'] = df['Date'].dt.year
df['Profit'] = df['Sales'] - df['cogs']
df
```

8) Create a new column called satisfied based on rating.
- Satisfied rule: rating >= 7 → Satisfied Else Not Satisfied

```python
df['satisfied'] = df['Rating'].apply(lambda x: True if x >= 7 else False)
df
```

1) Provide the total revenue generated by the supermarket.

```python
df['Sales'].sum().round(2)
```

2) Provide the total revenue by city and identify the city with the highest
revenue.

```python
city_sales = (
    df.groupby('City')['Sales']
    .sum().round(2).sort_values(ascending=False)
)
print(city_sales)
city_sales.index[0]
```

3) Provide the total profit by branch and identify the most profitable
branch.

```python
branch_profit = (
    df.groupby('Branch')['Profit']
    .sum().round(2).sort_values(ascending=False)
)
print(branch_profit)
branch_profit.index[0]
```

4) Provide the total revenue and profit by product category and identify
the top-performing category.

```python
product_sales = (
    df.groupby('Product line')[['Sales', 'Profit']]
    .sum().round(2).sort_values(by='Sales', ascending=False)
)
print(product_sales)
product_sales.index[0]
```

5) Provide the total spending by customer type and identify which
customer type spends more.

```python
customer_sales = (
    df.groupby('Customer type')['Sales']
    .sum().round(2).sort_values(ascending=False)
)
print(customer_sales)
customer_sales.index[0]
```

6) Provide the number of transactions by payment method and identify
the most popular payment method.

```python
customer_sales = (
    df.groupby('Payment')['Payment']
    .count()
)
print(customer_sales)
customer_sales.index[0]
```

7) Provide the average transaction value for the supermarket.

```python
df['Sales'].mean().round(2)
```

8) Provide the customer satisfaction percentage by branch and identify
the branch with the highest satisfaction.

```python
branch_satisfaction = (
    df.groupby('Branch')['satisfied']
    .mean().mul(100)
    .round(2)
    .sort_values(ascending=False)
)

print(branch_satisfaction)
print(branch_satisfaction.index[0])
```

9) Provide the total revenue by day and by month and identify the day
and month with the highest sales.

```python
month_day_sales = (
    df.groupby(['month', 'day'])['Sales']
    .sum().round(2).sort_values(ascending=False)
)
print(month_day_sales)
print(f"Month: {month_day_sales.index[0][0]}, Day: {month_day_sales.index[0][1]}")
```

10) Provide the overall percentage of satisfied customers/transactions.

```python
print(f"Satisfaction Rate: {df['satisfied'].mean() * 100:.2f}%")
```
